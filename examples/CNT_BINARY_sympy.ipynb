```python
# --- Symbols ---
J, K = sp.symbols('J K', positive=True)
DeltaG = sp.Symbol('DeltaG')                  # free-energy barrier [J]
T, kB = sp.symbols('T kB', positive=True)     # temperature [K], Boltzmann const [J/K]
gamma = sp.Symbol('gamma', positive=True)     # surface tension [N/m]
rc = sp.Symbol('rc', positive=True)           # critical radius [m]
Psat, Pc = sp.symbols('Psat Pc', positive=True)     # pressures [Pa]
DeltaP = sp.Symbol('DeltaP', positive=True)         # liquid tension [Pa]

rho_nL = sp.Symbol('rho_nL', positive=True)   # number density [1/m^3]
m_CO2 = sp.Symbol('m_CO2', positive=True)     # single-molecule mass of CO2 [kg]
N_c = sp.symbols('N_c', integer=True, positive=True)
i = sp.symbols('i', integer=True)
x = sp.IndexedBase('x')                       # mole fractions (–)
M = sp.IndexedBase('M')                       # molar masses [kg/mol]
Avogadro = sp.Symbol('Avogadro', positive=True)  # Avogadro constant [1/mol]
m_eff_bubble = sp.Symbol('m_eff_bubble', positive=True)

# --- Base equations (18)–(21) ---
eq_J = sp.Eq(J, K*sp.exp(-DeltaG/(kB*T)), evaluate=False)
eq_free_energy = sp.Eq(DeltaG, 4*sp.pi*gamma*rc**2/3, evaluate=False)
eq_liquid_tension = sp.Eq(DeltaP, Psat - Pc, evaluate=False)
eq_critical_radius = sp.Eq(rc, 2*gamma/(Psat - Pc), evaluate=False)

# --- K for pure CO2 and for mixture ---
K_pure_expr = rho_nL * sp.sqrt((2*gamma)/(sp.pi*m_CO2))  # pure CO2
# For mixture: 1/sqrt(m_eff) = Sum( x_i * sqrt(Avogadro / M_i) )
m_eff_inv_sqrt = sp.Sum(x[i] * sp.sqrt(Avogadro/M[i]), (i, 1, N_c))
K_mix_expr = rho_nL * sp.sqrt(2*gamma/sp.pi) * m_eff_inv_sqrt

# --- Fully substitute DeltaG and rc into J (style with Psat - Pc) ---
DeltaG_sub_ps = sp.simplify(eq_free_energy.rhs.subs(rc, eq_critical_radius.rhs))  # 4πγ (2γ/(Psat-Pc))^2 / 3
J_pure_full_ps = sp.Eq(J, K_pure_expr * sp.exp(-DeltaG_sub_ps/(kB*T)), evaluate=False)
J_mix_full_ps  = sp.Eq(J, K_mix_expr  * sp.exp(-DeltaG_sub_ps/(kB*T)), evaluate=False)

# --- Same but replace (Psat - Pc) by DeltaP using eq_liquid_tension ---
DeltaG_sub_dp = sp.simplify(DeltaG_sub_ps.subs(Psat - Pc, DeltaP))                # 4πγ (2γ/ΔP)^2 / 3
J_pure_full_dp = sp.Eq(J, K_pure_expr * sp.exp(-DeltaG_sub_dp/(kB*T)), evaluate=False)
J_mix_full_dp  = sp.Eq(J, K_mix_expr  * sp.exp(-DeltaG_sub_dp/(kB*T)), evaluate=False)

J_pure_full_ps
```

```python
# Mixture useage example

import math
from typing import Sequence
from scipy.constants import Avogadro as N_A, Boltzmann as kB

def K_mix_numeric(rho_L: float, M_list: Sequence[float], x_list: Sequence[float], gamma: float) -> float:
    """
    Pre-exponential factor K for a mixture (1/m^3·s).
    rho_L: mass density of liquid [kg/m^3]
    M_list: molar masses [kg/mol]
    x_list: mole fractions (sum to 1)
    gamma: surface tension [N/m]
    """
    assert abs(sum(x_list) - 1.0) < 1e-12, "x_list must sum to 1."
    M_mix = sum(xi * Mi for xi, Mi in zip(x_list, M_list))       # kg/mol
    rho_nL = (rho_L / M_mix) * N_A                               # 1/m^3
    sum_term = sum(xi * math.sqrt(N_A / Mi) for xi, Mi in zip(x_list, M_list))
    return rho_nL * math.sqrt(2 * gamma / math.pi) * sum_term

def nucleation_from_K(Jcrit: float, Psat: float, T: float, gamma: float, K: float):
    """
    Returns (DeltaP [Pa], Pc [Pa], rc [m]) given Jcrit, Psat, T, gamma, and K.
    """
    A = 16 * math.pi * gamma**3 / (3 * kB * T)                   # J·m / m^2 ... consistent in Pa with ΔP
    DeltaP = math.sqrt(A / math.log(K / Jcrit))                  # Pa    (requires K > Jcrit)
    Pc = Psat - DeltaP                                           # Pa
    rc = 2 * gamma / DeltaP                                      # m
    return DeltaP, Pc, rc

# ---------------------------
# EXAMPLE USAGE (edit values)
# ---------------------------
Jcrit = 1e12                           # [1/m^3·s]
Psat  = 29.816609*1e5                  # [Pa]
T     = 265.0                          # [K]
gamma = 5.75184769*1e-3                # [N/m]
rho_L = 964.482977                     # [kg/m^3]  (mixture liquid density)

# Mixture: e.g., CO2 + CH4
M_CO2 = 44.00985e-3                    # [kg/mol]
M_CH4 = 16.043e-3                      # [kg/mol]
x_CO2, x_CH4 = 0.992989, 0.007011      # mole fractions (must sum to 1)

K = K_mix_numeric(rho_L, [M_CO2, M_CH4], [x_CO2, x_CH4], gamma)
DeltaP, Pc, rc = nucleation_from_K(Jcrit, Psat, T, gamma, K)

# print(f"K = {K:.6e}  [1/m^3·s]")
# print(f"Delta-P = {DeltaP:.6e} Pa  ({DeltaP/1e5:.3f} bar)")
print(f"SHL (Pc) = {Pc:.6e} Pa  ({Pc/1e5:.3f} bar)")
# print(f"r_c = {rc:.6e} m  ({rc*1e9:.3f} nm)")
```

```python
import math
from typing import Sequence
from scipy.constants import Avogadro as N_A, Boltzmann as kB

def m_eff_droplet(M_list: Sequence[float], y_list: Sequence[float]) -> float:
    """
    Effective molecular mass (per molecule) for droplet nucleation in a mixture.
    sqrt(1/m_eff) = sum_i y_i / sqrt(m_i), where m_i = M_i / N_A.
    Returns m_eff in kg (per molecule).
    """
    assert abs(sum(y_list) - 1.0) < 1e-12, "y_list must sum to 1."
    inv_sqrt_m_eff = sum(y_i * math.sqrt(N_A / M_i) for y_i, M_i in zip(y_list, M_list))
    return 1.0 / (inv_sqrt_m_eff ** 2)

def rho_number_from_mass_density(rho_mass: float, M_list: Sequence[float], z_list: Sequence[float]) -> float:
    """
    Convert mixture mass density [kg/m^3] to number density [1/m^3]:
        rho_n = (rho_mass / M_mix) * N_A,  M_mix = sum z_i M_i
    z_list: phase composition (x for liquid, y for vapor).
    """
    M_mix = sum(zi * Mi for zi, Mi in zip(z_list, M_list))
    return (rho_mass / M_mix) * N_A

def K_scl_mix_numeric(
    rho_L: float,             # liquid mass density [kg/m^3]
    rho_g: float,             # vapor mass density [kg/m^3]
    M_list: Sequence[float],  # molar masses [kg/mol]
    x_list: Sequence[float],  # liquid mole fractions (sum=1)
    y_list: Sequence[float],  # vapor mole fractions (sum=1)
    gamma: float              # surface tension [N/m]
) -> float:
    """
    Pre-exponential factor K for SCL (droplet nucleation) in a mixture (1/m^3·s).
    K = (rho_nV^2 / rho_nL) * sqrt(2*gamma / (pi*m_eff_droplet))
    """
    rho_nL = rho_number_from_mass_density(rho_L, M_list, x_list)
    rho_nV = rho_number_from_mass_density(rho_g, M_list, y_list)
    m_eff  = m_eff_droplet(M_list, y_list)
    return (rho_nV**2 / rho_nL) * math.sqrt(2.0 * gamma / (math.pi * m_eff))

def scl_from_K(Jcrit: float, Psat: float, T: float, gamma: float, K: float, rho_nL: float):
    """
    Returns (Pc [Pa], rc [m], lnS) for SCL given Jcrit, Psat, T, gamma, K, and rho_nL.
    """
    A_over_kBT = 16.0 * math.pi * gamma**3 / (3.0 * (rho_nL**2) * (kB*T)**3)
    lnS = math.sqrt(A_over_kBT / math.log(K / Jcrit))
    Pc  = Psat * math.exp(lnS)
    rc  = 2.0 * gamma / (rho_nL * kB * T * lnS)
    return Pc, rc, lnS

# ---------------------------
# EXAMPLE USAGE (SCL only)
# ---------------------------
if __name__ == "__main__":
    # Known parameters
    Jcrit = 1e12                  # [1/m^3·s]
    Psat  = 40.728415e5           # [Pa]
    T     = 265.0                 # [K]
    gamma = 4.26321575e-3         # [N/m]
    rho_L = 908.919221            # [kg/m^3]

    # Vapor phase for SCL
    rho_g = 98.57181                   # [kg/m^3]  <-- placeholder, replace with your EoS value
    y_CO2, y_CH4 = 0.754938, 0.245062

    # Liquid composition (same as SHL)
    x_CO2, x_CH4 = 0.992989, 0.007011

    # Components
    M_CO2 = 44.00985e-3           # [kg/mol]
    M_CH4 = 16.043e-3             # [kg/mol]

    M_list  = [M_CO2, M_CH4]
    x_list  = [x_CO2, x_CH4]
    y_list  = [y_CO2, y_CH4]

    # Compute prefactor
    K_SCL = K_scl_mix_numeric(rho_L, rho_g, M_list, x_list, y_list, gamma)
    rho_nL = rho_number_from_mass_density(rho_L, M_list, x_list)

    # Compute Pc, rc, lnS
    Pc_SCL, rc_SCL, lnS = scl_from_K(Jcrit, Psat, T, gamma, K_SCL, rho_nL)

    # Print results
    print(f"SCL (Pc) = {Pc_SCL:.6e} Pa  ({Pc_SCL/1e5:.3f} bar)")
    # print(f"SCL (rc) = {rc_SCL:.6e} m  ({rc_SCL*1e9:.3f} nm)")
    # print(f"ln S = {lnS:.6e}")
    # print(f"K_SCL = {K_SCL:.6e} [1/m^3·s]")
```

```python
import math
from typing import Sequence
from scipy.constants import Avogadro as N_A, Boltzmann as kB

def m_eff_droplet(M_list: Sequence[float], y_list: Sequence[float]) -> float:
    """
    Effective molecular mass (per molecule) for droplet nucleation in a mixture.
    sqrt(1/m_eff) = sum_i y_i / sqrt(m_i), where m_i = M_i / N_A.
    Returns m_eff in kg (per molecule).
    """
    assert abs(sum(y_list) - 1.0) < 1e-12, "y_list must sum to 1."
    inv_sqrt_m_eff = sum(y_i * math.sqrt(N_A / M_i) for y_i, M_i in zip(y_list, M_list))
    return 1.0 / (inv_sqrt_m_eff ** 2)

def rho_number_from_mass_density(rho_mass: float, M_list: Sequence[float], z_list: Sequence[float]) -> float:
    """
    Convert mixture mass density [kg/m^3] to number density [1/m^3]:
        rho_n = (rho_mass / M_mix) * N_A,  M_mix = sum z_i M_i
    z_list: phase composition (x for liquid, y for vapor).
    """
    M_mix = sum(zi * Mi for zi, Mi in zip(z_list, M_list))
    return (rho_mass / M_mix) * N_A

def K_scl_mix_numeric(
    rho_L: float,             # liquid mass density [kg/m^3]
    rho_g: float,             # vapor mass density [kg/m^3]
    M_list: Sequence[float],  # molar masses [kg/mol]
    x_list: Sequence[float],  # liquid mole fractions (sum=1)
    y_list: Sequence[float],  # vapor mole fractions (sum=1)
    gamma: float              # surface tension [N/m]
) -> float:
    """
    Pre-exponential factor K for SCL (droplet nucleation) in a mixture (1/m^3·s).
    K = (rho_nV^2 / rho_nL) * sqrt(2*gamma / (pi*m_eff_droplet))
    """
    rho_nL = rho_number_from_mass_density(rho_L, M_list, x_list)
    rho_nV = rho_number_from_mass_density(rho_g, M_list, y_list)
    m_eff  = m_eff_droplet(M_list, y_list)
    return (rho_nV**2 / rho_nL) * math.sqrt(2.0 * gamma / (math.pi * m_eff))

def scl_from_K(Jcrit: float, Psat: float, T: float, gamma: float, K: float, rho_nL: float):
    """
    Returns (Pc [Pa], rc [m], lnS) for SCL given Jcrit, Psat, T, gamma, K, and rho_nL.
    """
    A_over_kBT = 16.0 * math.pi * gamma**3 / (3.0 * (rho_nL**2) * (kB*T)**3)
    lnS = math.sqrt(A_over_kBT / math.log(K / Jcrit))
    Pc  = Psat * math.exp(lnS)
    rc  = 2.0 * gamma / (rho_nL * kB * T * lnS)
    return Pc, rc, lnS

# ---------------------------
# EXAMPLE USAGE (SCL only)
# ---------------------------
if __name__ == "__main__":
    # Known parameters
    Jcrit = 1e12                  # [1/m^3·s]
    Psat  = 18.48807e5           # [Pa]
    T     = 220                 # [K]
    gamma = 14.07199084e-3         # [N/m]
    rho_L = 1168.900069            # [kg/m^3]

    # Vapor phase for SCL
    rho_g = 31.488815                   # [kg/m^3]  <-- placeholder, replace with your EoS value
    y_CO2, y_CH4 = 0.379782, 0.620218

    # Liquid composition (same as SHL)
    x_CO2, x_CH4 = 1, 0

    # Components
    M_CO2 = 44.00985e-3           # [kg/mol]
    M_CH4 = 16.043e-3             # [kg/mol]

    M_list  = [M_CO2, M_CH4]
    x_list  = [x_CO2, x_CH4]
    y_list  = [y_CO2, y_CH4]

    # Compute prefactor
    K_SCL = K_scl_mix_numeric(rho_L, rho_g, M_list, x_list, y_list, gamma)
    rho_nL = rho_number_from_mass_density(rho_L, M_list, x_list)

    # Compute Pc, rc, lnS
    Pc_SCL, rc_SCL, lnS = scl_from_K(Jcrit, Psat, T, gamma, K_SCL, rho_nL)

    # Print results
    print(f"SCL (Pc) = {Pc_SCL:.6e} Pa  ({Pc_SCL/1e5:.3f} bar)")
    # print(f"SCL (rc) = {rc_SCL:.6e} m  ({rc_SCL*1e9:.3f} nm)")
    # print(f"ln S = {lnS:.6e}")
    # print(f"K_SCL = {K_SCL:.6e} [1/m^3·s]")
```

```python
def K_mix_numeric(rho_L: float, M_list: Sequence[float], x_list: Sequence[float], gamma: float) -> float:
    """
    Pre-exponential factor K for a mixture (1/m^3·s).
    rho_L: mass density of liquid [kg/m^3]
    M_list: molar masses [kg/mol]
    x_list: mole fractions (sum to 1)
    gamma: surface tension [N/m]
    """
    assert abs(sum(x_list) - 1.0) < 1e-12, "x_list must sum to 1."
    M_mix = sum(xi * Mi for xi, Mi in zip(x_list, M_list))       # kg/mol
    rho_nL = (rho_L / M_mix) * N_A                               # 1/m^3
    sum_term = sum(xi * math.sqrt(N_A / Mi) for xi, Mi in zip(x_list, M_list))
    return rho_nL * math.sqrt(2 * gamma / math.pi) * sum_term

def nucleation_from_K(Jcrit: float, Psat: float, T: float, gamma: float, K: float):
    """
    Returns (DeltaP [Pa], Pc [Pa], rc [m]) given Jcrit, Psat, T, gamma, and K.
    """
    A = 16 * math.pi * gamma**3 / (3 * kB * T)                   # J·m / m^2 ... consistent in Pa with ΔP
    DeltaP = math.sqrt(A / math.log(K / Jcrit))                  # Pa    (requires K > Jcrit)
    Pc = Psat - DeltaP                                           # Pa
    rc = 2 * gamma / DeltaP                                      # m
    return DeltaP, Pc, rc

# ---------------------------
# EXAMPLE USAGE (edit values)
# ---------------------------
Jcrit = 1e12                           # [1/m^3·s]
Psat  = 34.293136e5                    # [Pa]
T     = 270.0                          # [K]
gamma = 0.0047784                      # [N/m]
rho_L = 935.18                     # [kg/m^3]  (mixture liquid density)

# Mixture: e.g., CO2 + CH4
M_CO2 = 44.00985e-3                    # [kg/mol]
M_CH4 = 16.043e-3                      # [kg/mol]
x_CO2, x_CH4 = 0.99, 0.01                # mole fractions (must sum to 1)

K = K_mix_numeric(rho_L, [M_CO2, M_CH4], [x_CO2, x_CH4], gamma)
DeltaP, Pc, rc = nucleation_from_K(Jcrit, Psat, T, gamma, K)

# print(f"K = {K:.6e}  [1/m^3·s]")
# print(f"Delta-P = {DeltaP:.6e} Pa  ({DeltaP/1e5:.3f} bar)")
print(f"SHL (Pc) = {Pc:.6e} Pa  ({Pc/1e5:.3f} bar)")
# print(f"r_c = {rc:.6e} m  ({rc*1e9:.3f} nm)")
```
